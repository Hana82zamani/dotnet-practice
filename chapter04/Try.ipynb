```
static bool TryCreateUser(string? name, out string result)
{
    if (string.IsNullOrWhiteSpace(name))
    {
        result = default!;
        return false;
    }
    result= $"Hello {name}";
    return true;
}
if(TryCreateUser("Hana",out string message))
{
    Console.WriteLine(message);
}
else
{
    Console.WriteLine("invalid name");

}
```
