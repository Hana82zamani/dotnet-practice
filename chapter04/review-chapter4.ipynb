```
static double CalculateArea(int side)
{
    return side * side;
}

static int CalculateArea(int length, int width)
{
    return length * width;
}

static double CalculateArea(double radius)
{
    return 3.14 * radius * radius;
}


Console.WriteLine(CalculateArea(5));
Console.WriteLine(CalculateArea(5,10));
Console.WriteLine(CalculateArea(3.5));
/// (چندتابع با نام یکسان و پارمتر های متفاوت) Method Overloading حل تمرین تابع مدل 
```

```
static string Convert(int number)
{
    return $"{number}";

}

static string Convert(double number)
{
    return number.ToString("F2");
}

static string Convert(string text, int repeat)
{
    string result= "" ;

    for(int i=1;i<=repeat;i++)
    {
    result=result+text;
    }
    return result;
} 
Console.WriteLine(Convert(5));
Console.WriteLine(Convert(3.15248));
Console.WriteLine(Convert("Hi",5));
///و شرط حلقه از Overloading تمرین تابع 
```

```
static void ProcessOrder(decimal amount)
{
    if (amount <= 0)
    {
        throw new ArgumentOutOfRangeException("amount shoudn't be negative");

    }
    else if (amount > 100000)
    {
        throw new InvalidOperationException("amount is greater than 100000");
    }
    else
    {
        Console.WriteLine("Order processed successfully.");
    }
}


static void checkout()
{

try
{
    ProcessOrder(-5);
}
catch(Exception ex)
{
    Console.WriteLine(ex.Message);

    throw; // بهم نریزه از استفاده میشهcall stack برای اینکه ترتیب 
}

}
checkout();
```
